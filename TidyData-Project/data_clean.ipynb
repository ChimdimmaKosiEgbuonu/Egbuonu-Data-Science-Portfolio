```python
#loading the necessary libraries

import pandas as pd            
import seaborn as sns          
import matplotlib.pyplot as plt  


#reads the csv file
df = pd.read_csv("mutant_moneyball.csv")

#gives general overview of the data
df.describe()
```

```python
df.info() #gives info on the dataset
```

```python
print(df.columns.tolist()) #lists the columns in the dataset
```

```python
df_cash = pd.melt(
        df,
        id_vars = ["Member"],
        var_name = "Category",
        value_name = "Total_Amount")

print("\nMelted (Tidy) Mutant Moneyball DataFrame:")
print(df_cash) #melts the data
```

```python

#Note: I used Generative AI in assistance for this specific block of code

df_cash['Category'] = df_cash['Category'].str.replace('TotalValue', '', regex=False) # removes the "TotalValue" part from "Category"

df_cash[['Decade', "Origin"]] = df_cash['Category'].str.split('_', expand=True) #splits "decade" and "origin" away from "Category" into their own columns 

df_cash = df_cash.drop(columns=['Category']) #removes the category column all together, since it's now been split

df_cash['Total_Amount'] = (
    df_cash['Total_Amount']
    .astype(str)
    .str.strip()                       
    .str.replace('$', '', regex=False) 
    .str.replace(',', '', regex=False) # this block of code converts all the numbers to be the same format, without dollar signs and commas
)
df_cash['Total_Amount'] = pd.to_numeric(df_cash['Total_Amount'], errors='coerce')

print(df_cash) #returns the new data
```

```python
ax = sns.barplot(data=df_cash, x= "Origin", y ="Total_Amount", hue = "Origin",  palette ='magma') #barplot demonstrating the total amount based on origin
ax.set_title("Barplot Showing Total Amount Based on Origin") #sets title of boxplot
ax.set_yscale("log") #sets the range of numbers to be in log form
```

```python
ax = sns.boxplot(data= df_cash, x = "Decade", y= "Total_Amount", hue = "Decade", palette ='mako') #boxplot demonstrating the total amount based on the decade
ax.set_yscale("log") #sets the range of numbers to be in log form
```

```python
# pivot table that shows the total amount based on decade and origin for each member

#replaces the null values with zero
df_cash = df_cash.fillna(0)

#returns the pivot table
pd.pivot_table(
    df_cash,
    values = "Total_Amount",
    index = ["Decade", "Origin"],
    columns = "Member",
    aggfunc = "mean"
)
```
